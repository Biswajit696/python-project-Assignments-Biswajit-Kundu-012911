```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference:", A.difference(B))
```

```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

A = set(list1)
B = set(list2)

common = A.intersection(B)

print("Common Elements:", common)
```

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print("A is Subset of B:", A.issubset(B))
print("B is Superset of A:", B.issuperset(A))
```

```python
python_students = {"Aman", "Riya", "Soham", "Priya"}
java_students = {"Riya", "Priya", "Rahul", "Ankit"}

result = python_students.symmetric_difference(java_students)

print("Students enrolled in exactly one course:")
print(result)
```
