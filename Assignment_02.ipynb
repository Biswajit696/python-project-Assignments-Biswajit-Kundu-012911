```python
n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1):
    sum = sum + i ** 2

print("Sum of the series =", sum)
```

```python
n = int(input("Enter a number: "))
count = 0

if n <= 1:
    print(n, "is not a Prime Number")
else:
    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1

    if count == 2:
        print(n, "is a Prime Number")
    else:
        print(n, "is not a Prime Number")
```
