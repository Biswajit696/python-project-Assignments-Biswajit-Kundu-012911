```python
numbers = []

for i in range(10):
    n = int(input("Enter a number: "))
    numbers.append(n)

largest = numbers[0]
smallest = numbers[0]

for i in numbers:
    if i > largest:
        largest = i
    if i < smallest:
        smallest = i

print("Largest number =", largest)
print("Smallest number =", smallest)
```

```python
numbers = [10, 20, 10, 30, 20, 40]
unique = []

for i in numbers:
    if i not in unique:
        unique.append(i)

print("Original List:", numbers)
print("List without duplicates:", unique)
```

```python
numbers = [12, 7, 5, 18, 20, 9]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even:", even)
print("Odd:", odd)
```
