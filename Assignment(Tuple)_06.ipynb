```python
t = (1, 2, 2, 3, 1, 4, 3)

unique = ()

for i in t:
    if i not in unique:
        unique = unique + (i,)

print("Original Tuple:", t)
print("Unique Tuple:", unique)
```

```python
students = (
    ("Riya", 85),
    ("Aman", 92),
    ("Soham", 78),
    ("Priya", 89)
)

sorted_students = sorted(students, key=lambda x: x[1], reverse=True)

print("Students in Descending Order:")

for name, marks in sorted_students:
    print(name, marks)
```

```python
numbers = []

for i in range(5):
    n = int(input("Enter a number: "))
    numbers.append(n)

t = tuple(numbers)

total = sum(t)
average = total / 5

print("Tuple:", t)
print("Sum =", total)
print("Average =", average)3
```

```python
t = (10, 20, 10, 30, 10, 40, 20)

count = t.count(10)

print("Number 10 occurs", count, "times")
```

```python
t = (10, 25, 5, 40, 15, 30)

largest = t[0]
smallest = t[0]

for i in t:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("Largest =", largest)
print("Smallest =", smallest)
```
