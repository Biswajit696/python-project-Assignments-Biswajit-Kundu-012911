```python
n = int(input("Enter a number: "))
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit
    n = n // 10

print("Sum of digits =", sum)
```

```python
n = int(input("Enter a number: "))
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reversed number =", reverse)
```

```python
sum = 0

n = int(input("Enter a number (0 to stop): "))

while n != 0:
    sum = sum + n
    n = int(input("Enter a number (0 to stop): "))

print("Sum of all numbers =", sum)
```
